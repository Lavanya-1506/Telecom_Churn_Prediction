```python
import pandas as pd

df = pd.read_csv("../../data/raw/teleco_churn.csv")
df.head()
```

```python
# Churn Distribution
import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(
    x="Churn",
    data=df
)

plt.show()
```

```python

# Contract Analysis
sns.countplot(
    x="Contract",
    hue="Churn",
    data=df
)

plt.xticks(rotation=20)

plt.show()
```

```python
# Monthly Charges

sns.boxplot(
    x="Churn",
    y="MonthlyCharges",
    data=df
)
```

```python
# Tenure Distribution
sns.histplot(
    data=df,
    x="tenure",
    hue="Churn",
    kde=True
)
```

```python
# Correlation Heatmap

corr_matrix = df.corr(numeric_only=True)

sns.heatmap(
    corr_matrix,
    annot=True
)
```
