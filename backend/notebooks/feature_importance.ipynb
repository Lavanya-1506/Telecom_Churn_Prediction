```python
import pandas as pd
import joblib

df = pd.read_csv("../../data/raw/WA_Fn-UseC_-Telco-Customer-Churn.csv")
df.head()
```

```python
X = df.drop("Churn", axis=1)
```

```python
import joblib

lr = joblib.load("../models/best_model.pkl")
```

```python
print("Number of Features:", len(X.columns))
print("Number of Coefficients:", len(lr.coef_[0]))
```

```python
print("X shape:", X.shape)
print("Model shape:", lr.coef_.shape)

for i, col in enumerate(X.columns):
    print(i, col)
```

```python
len(X.columns)
```

```python
X_temp = X.drop(
    columns=[
        "AvgRevenue",
        "TenureGroup_1-2 Years",
        "TenureGroup_2-4 Years",
        "TenureGroup_4-6 Years"
    ]
)

print(X_temp.shape)
```

```python
coefs = lr.coef_.ravel()

model_feature_names = getattr(lr, "feature_names_in_", None)
features = list(model_feature_names) if model_feature_names is not None else list(X_temp.columns)

if len(features) != len(coefs):
    min_len = min(len(features), len(coefs))
    features = features[:min_len]
    coefs = coefs[:min_len]

importance = pd.DataFrame({"Feature": features, "Coefficient": coefs})
importance = importance.sort_values(by="Coefficient", ascending=False)
importance.head(15)
```
